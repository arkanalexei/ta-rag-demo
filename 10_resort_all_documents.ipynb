```python
CUTOFF = 22780
```

```python
from utils.utils import load_from_pickle, save_to_pickle

def split_into_batches(input_list, n):
    batch_size = len(input_list) // n
    remainder = len(input_list) % n

    batches = []
    start = 0

    for i in range(n):
        end = start + batch_size + (1 if remainder > 0 else 0)
        batches.append(input_list[start:end])
        start = end
        remainder -= 1

    return batches

all_documents = load_from_pickle('all_documents')
all_documents_batches = split_into_batches(all_documents, len(all_documents))
```

```python
cutoff_documents_batches = all_documents_batches[CUTOFF:]
```

```python
print(len(all_documents_batches), len(cutoff_documents_batches))
print(len(cutoff_documents_batches) + CUTOFF)
```

```python
from utils.utils import clean_string

tentang_dict = {}
for doc in cutoff_documents_batches:
    metadata = clean_string(doc[0].metadata['tentang'])
    if metadata in tentang_dict.keys():
        tentang_dict[metadata] += 1
    else:
        tentang_dict[metadata] = 1

tentang_dict = dict(sorted(tentang_dict.items(), key=lambda item: item[1], reverse=True))
```

```python
tentang_dict
```

```python
from collections import defaultdict

tentang_to_docs = defaultdict(list)
for doc in cutoff_documents_batches:
    metadata = clean_string(doc[0].metadata['tentang'])
    tentang_to_docs[metadata].append(doc[0])
```

```python
import random

sorted_cutoff_documents = []
for tentang, count in tentang_dict.items():
    docs = tentang_to_docs[tentang]
    random.shuffle(docs)  # Randomize order within the 'tentang' group
    sorted_cutoff_documents.extend(docs)
```
