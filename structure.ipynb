```python
import os

EXCLUDE_DIRS = {"venv", "env", "env2"}  # Directories to exclude

def print_tree(root_dir, indent="", max_files=25):
    """
    Recursively prints the directory and file structure, with a limit on the number of files displayed per folder.
    Avoids descending into specific excluded directories.

    :param root_dir: The root directory to start scanning.
    :param indent: Indentation for the current level in the tree.
    :param max_files: Maximum number of files to display per folder.
    """
    items = os.listdir(root_dir)
    dir_items = []
    file_items = []

    # Separate directories and files
    for item in items:
        item_path = os.path.join(root_dir, item)
        if os.path.isdir(item_path):
            if item in EXCLUDE_DIRS:
                continue  # Skip excluded directories
            dir_items.append(item)
        else:
            file_items.append(item)

    # Print files in the root directory
    if indent == "":
        if len(file_items) > max_files:
            for file in file_items[:max_files]:
                print(f"{indent}📄 {file}")
            print(f"{indent}... ({len(file_items) - max_files} more files)")
        else:
            for file in file_items:
                print(f"{indent}📄 {file}")

    # Print directories
    for directory in dir_items:
        print(f"{indent}📁 {directory}/")
        print_tree(os.path.join(root_dir, directory), indent + "    ", max_files)

# Start printing the directory structure from the current directory
if __name__ == "__main__":
    print_tree(os.getcwd())
```
