```python
def generate_readable_content(metadata):
    """
    Generates a natural language string from metadata dictionary for use in page_content.
    """
    readable_content = [
        f"Here are informations of {metadata['label'].split('/')[-1].replace('_', ' ')} regulation:"
    ]
    
    # Format key metadata fields with more descriptive language
    readable_content.append(f"Identifier of the regulation is {metadata['id'].split('/')[-1]}.")
    readable_content.append(f"Label or name of the regulation is {metadata['label'].split('/')[-1].replace('_', ' ')}.")
    
    if metadata['description'] != "N/A":
        readable_content.append(f"Here is the description of the regulation: {metadata['description']}.")

    # Convert list fields into readable sentences
    if metadata['has_content']:
        contents = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['has_content']])
        readable_content.append(f"The regulation contains the following chapters or parts: {contents}.")

    if metadata['has_creator']:
        creators = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['has_creator']])
        readable_content.append(f"The regulation was created by: {creators}.")

    if metadata['considers']:
        considerations = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['considers']])
        readable_content.append(f"The regulation considers the following: {considerations}.")

    if metadata['has_legal_basis']:
        legal_bases = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['has_legal_basis']])
        readable_content.append(f"The regulation has the following legal bases: {legal_bases}.")

    if metadata['implements']:
        implementations = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['implements']])
        readable_content.append(f"The regulation implements the following regulations: {implementations}.")

    if metadata['repeals']:
        repealed_items = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['repeals']])
        readable_content.append(f"The regulation repeals the following: {repealed_items}.")

    if metadata['amends']:
        amendments = ", ".join([item.split('/')[-1].replace('_', ' ') for item in metadata['amends']])
        readable_content.append(f"The regulation amends the following: {amendments}.")

    # Add the remaining single attributes in human-readable format
    if metadata.get('regulation_number') != "N/A":
        readable_content.append(f"The regulation number of the regulation is {metadata['regulation_number']}.")

    if metadata.get('regulation_year') != "N/A":
        readable_content.append(f"The regulation was issued in the year {metadata['regulation_year']}.")

    if metadata.get('promulgation_place') != "N/A":
        readable_content.append(f"The place of promulgation for the regulation is {metadata['promulgation_place']}.")

    if metadata.get('enaction_date') != "N/A":
        readable_content.append(f"The date of enactment for the regulation is {metadata['enaction_date']}.")

    if metadata.get('enaction_location') != "N/A":
        readable_content.append(f"The location of enactment for the regulation is {metadata['enaction_location']}.")

    if metadata.get('promulgation_date') != "N/A":
        readable_content.append(f"The regulation was promulgated on {metadata['promulgation_date']}.")

    if metadata.get('promulgation_location') != "N/A":
        readable_content.append(f"The location of promulgation for the regulation is {metadata['promulgation_location']}.")

    if metadata.get('promulgation_office') != "N/A":
        readable_content.append(f"The office responsible for promulgation is {metadata['promulgation_office']}.")

    if metadata.get('promulgation_official') != "N/A":
        readable_content.append(f"The official responsible for promulgation is {metadata['promulgation_official']}.")

    # Combine readable content into a single string
    return "\n".join(readable_content)
```

```python
# Recursive function to gather all descriptions at Section and Item level, with cycle detection
def gather_descriptions(node, visited):
    if node in visited:
        return []  # Avoid infinite recursion by returning an empty list if the node is already visited

    visited.add(node)  # Mark this node as visited
    descriptions = []
    
    # If the node has dct:description, add it to descriptions
    for description in g.objects(node, DCTERMS.description):
        clean_desc = str(description).replace("http://example.com/base/", "").replace("_", " ")
        descriptions.append(clean_desc)
    
    # Recursively follow lexid-s:hasPart to find more Section and Item nodes
    for part in g.objects(node, LEXID_S.hasPart):
        descriptions.extend(gather_descriptions(part, visited))
    
    visited.remove(node)  # Unmark this node when done with it
    return descriptions
```

```python
import os
import logging
from dotenv import load_dotenv
import nest_asyncio
import pandas as pd
from tqdm import tqdm
from ragas.testset.generator import TestsetGenerator
from ragas.testset.evolutions import simple, reasoning, multi_context, conditional

from utils.utils import load_from_pickle, save_to_pickle
from utils.llm import get_llm, get_embedding_model

# Setup loggingx
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

# Load environment variables and apply nest_asyncio
load_dotenv()
nest_asyncio.apply()

def split_into_batches(input_list, n):
    batch_size = len(input_list) // n
    remainder = len(input_list) % n

    batches = []
    start = 0

    for i in range(n):
        end = start + batch_size + (1 if remainder > 0 else 0)
        batches.append(input_list[start:end])
        start = end
        remainder -= 1

    return batches

def get_last_processed_batch():
    try:
        with open('pickles/progress-ttl-meta.log', 'r') as f:
            return int(f.read().strip())
    except FileNotFoundError:
        return 0

def save_progress(batch_num):
    with open('pickles/progress-ttl-meta.log', 'w') as f:
        f.write(str(batch_num))
        
```

```python
def main():
    # Load documents
    all_documents = load_from_pickle('all_documents_metadata_ttl')
    logging.info(f"Loaded {len(all_documents)} documents")

    # Ensure unique identifiers in metadata
    for i, doc in enumerate(all_documents):
        doc.metadata["unique_id"] = f"doc_{i}"

    # Split into batches of 1 item each
    all_documents_batches = split_into_batches(all_documents, len(all_documents)) # 857
    logging.info(f"Split into {len(all_documents_batches)} batches")

    # Setup TestsetGenerator
    generator_llm = get_llm("nvidia-llama3.1")
    critic_llm = get_llm("nvidia-llama3.1")
    embeddings = get_embedding_model("BAAI/bge-m3")
    generator = TestsetGenerator.from_langchain(generator_llm=generator_llm, critic_llm=critic_llm, embeddings=embeddings)

    # Verify documents are added to docstore
    logging.info(f"Number of documents in docstore: {len(generator.docstore.nodes)}")

    # Set test parameters
    test_size = 1
    distributions = {simple: 1}

    # Get the last processed batch
    last_processed_batch = get_last_processed_batch()
    logging.info(f"Resuming from batch {last_processed_batch}")

    # Load existing testset if available
    if last_processed_batch > 0:
        merged_testset = load_from_pickle(f'merged_testset_ttl_{last_processed_batch}')
        logging.info(f"Loaded existing testset with {len(merged_testset)} entries")
    else:
        merged_testset = pd.DataFrame()

      # Process batches
    for batch_index, batch in enumerate(tqdm(all_documents_batches[last_processed_batch:], initial=last_processed_batch, total=len(all_documents_batches))):
        # Calculate the actual batch number by adding the last processed batch index
        actual_batch_num = last_processed_batch + batch_index

        # print(f"Batch index: {batch_index}")
        # print(f"Batch: {batch}")
        # counter = 0
        try:
            # if counter == 4:
            #     break
            # counter += 1
            testset = generator.generate_with_langchain_docs(batch, test_size=test_size, distributions=distributions)
            batch_df = testset.to_pandas()
            merged_testset = pd.concat([merged_testset, batch_df], ignore_index=True)
            # break

            # Save progress every 5 batches
            if (actual_batch_num + 1) % 5 == 0:
                save_to_pickle(merged_testset, f'merged_testset_ttl_{actual_batch_num + 1}')
                save_progress(actual_batch_num + 1)
                logging.info(f"Saved progress: {actual_batch_num + 1}/{len(all_documents_batches)} batches processed")

        except Exception as e:
            logging.error(f"Error processing batch {actual_batch_num}: {str(e)}")
            # Save progress before exiting
            save_to_pickle(merged_testset, f'merged_testset_ttl_{actual_batch_num}')
            save_progress(actual_batch_num)
            raise

    # Save final result
    
    # save_to_pickle(merged_testset, 'final_merged_testset_ttl')
    save_to_pickle(merged_testset, 'final_merged_testset_meta_2_ttl')
    logging.info("Processing completed. Final testset saved as 'pickles/final_merged_testset_meta_2_ttl'")

if __name__ == "__main__":
    testing = main()
    print(testing)
```

### Question Cleaning

```python
import re

def extract_question(text):
    """
    Extracts just the question from a longer text that may include context, explanations,
    and formatting artifacts.
    
    Args:
        text (str): The input text containing a question and potentially other content
        
    Returns:
        str: The cleaned question text
    """
    # First, try to find a quoted question
    quote_patterns = [
        r'"([^"]+?\?)"',  # Double quotes
        r"'([^']+?\?)'",  # Single quotes
        r'"([^"]+?\?)(?!")',  # Double quote at start only
        r"'([^']+?\?)(?!')"   # Single quote at start only
    ]
    
    # Common intro phrases to remove
    intro_patterns = [
        r"Here is (?:the )?(?:rewritten )?question(?:[^:]*)?:\s*",
        r"This (?:revised )?question:\s*",
        r"Question:\s*"
    ]
    
    # Remove common intro phrases
    for pattern in intro_patterns:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE)
    
    # Try each quote pattern
    for pattern in quote_patterns:
        match = re.search(pattern, text)
        if match:
            return match.group(1).strip()
    
    # If no quotes found, try to find the first sentence that ends with a question mark
    match = re.search(r"([^.!?\n]+?\?)", text)
    if match:
        return match.group(1).strip()
    
    # If everything fails, return the first non-empty line that ends with a question mark
    for line in text.split('\n'):
        line = line.strip()
        if line.endswith('?'):
            return line
    
    # If no question mark is found, return the first non-empty line
    for line in text.split('\n'):
        line = line.strip()
        if line:
            return line
    
    return text.strip()

# Apply the function to the dataset
filtered_dataset['cleaned_question'] = filtered_dataset['question'].apply(extract_question)
```

### Dataset Validation to Testset

```python
import pandas as pd
import openai
import concurrent.futures
import time
from utils.utils import load_from_pickle, save_to_pickle

# Set your OpenAI API key
api_key = "" # Replace with your actual API key

client = openai.OpenAI(
    api_key=api_key,  # This is the default and can be omitted
)

# Load the DataFrame
data = load_from_pickle("filtered_testset_2_ttl")

# Function to construct the prompt and get the validation result
def get_validation(row):
    question = row['cleaned_question']
    context = ' '.join(row['contexts'])  # Combine context if it's a list
    ground_truth = row['ground_truth']
    
    # Construct a concise prompt
    prompt = f"""
Based on the context below, classify the answer as one of the choices (A, B, C, or D):

Context:
{context}

Question:
{question}

Answer:
{ground_truth}

Choices:
A. Correct answer (answers the question) and matches the given context
B. Correct answer (answers the question) but does NOT match the given context
C. Incorrect answer (does not answer the question) but the question matches the context
D. Incorrect answer (does not answer the question) and the question does NOT match the context

Provide only the letter corresponding to the correct choice (A, B, C, or D).
"""
    max_retries = 5
    retry_delay = 5  # seconds
    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                        model="gpt-4o-mini",
                        messages=[
                            {"role": "user", "content": prompt,}
                        ],
                        max_tokens=1,  # Limit to 1 token since the expected output is a single character
                        temperature=0.0  # Use deterministic output
                    )
            # response = openai.ChatCompletion.create(
            #     model="gpt-4o-mini",
            #     messages=[
            #         {"role": "user", "content": prompt}
            #     ],
            #     max_tokens=1,  # Limit to 1 token since the expected output is a single character
            #     temperature=0.0  # Use deterministic output
            # )
            # Extract the assistant's reply
            validation = response.choices[0].message.content.strip()
            return validation
        except openai.RateLimitError as e:
            if attempt < max_retries - 1:
                print(f"Rate limit reached. Retrying in {retry_delay} seconds...")
                time.sleep(retry_delay)
            else:
                print(f"Error processing row {row.name}: {e}")
                return None
        except Exception as e:
            print(f"Error processing row {row.name}: {e}")
            return None

# Apply the function to the DataFrame using parallel processing
def process_dataframe(df):
    results = []
    max_workers = 5  # Adjust this number based on your rate limit and performance needs
    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {executor.submit(get_validation, row): idx for idx, row in df.iterrows()}
        for future in concurrent.futures.as_completed(futures):
            idx = futures[future]
            try:
                validation = future.result()
            except Exception as e:
                print(f"Error at index {idx}: {e}")
                validation = None
            results.append((idx, validation))
    return results

# example_row = data.iloc[0]
# result = get_validation(example_row)

# Run the processing function
validation_results = process_dataframe(data)
validation_dict = dict(validation_results)
data['validate'] = data.index.map(validation_dict)

# Save the updated DataFrame
filename = "validated_graph_testset"
save_to_pickle(data, filename)

print(f"Validation complete. The updated DataFrame has been saved as '{filename}'.")
```
