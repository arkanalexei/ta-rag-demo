```python
from utils.utils import load_from_pickle

experiment = load_from_pickle("llm_experiments/baseline")
experiment
```

```python
import os

directory = "llm_experiments"
pkl_files = [f for f in os.listdir(f"pickles/{directory}") if f.endswith(".pkl")]

master_data = []

for file in pkl_files:
    experiment_name = file.replace(".pkl", "")
    file = file.split(".")[0]
    path = os.path.join(directory, file)
    data = load_from_pickle(path)  
    for key, row in data.items():
        row["Experiment"] = experiment_name
        master_data.append(row)
```

```python
import pandas as pd

df = pd.DataFrame(master_data)
len(df)
```

```python
validated_testset = load_from_pickle("validated_testset")
non_a_qa_pairs = validated_testset[validated_testset.validate != "A"]
non_a_questions = non_a_qa_pairs["cleaned_question"].tolist()
# non_a_questions
```

```python
filtered_df = df[~df['question'].isin(non_a_questions)]
len(filtered_df)
```

```python
# df
output_path = "experiments_results.xlsx"
filtered_df.to_excel(output_path, index=False, engine="openpyxl")
print(f"Data successfully exported to {output_path}")
```
