```python
## OPTIMIZED METHOD ##
## From 2 minutes to 2 seconds !!! ##

import pandas as pd
import re
import numpy as np

def clean_string_optimized(s):
    """
    Optimized string cleaning function that combines multiple operations.
    """
    # Compile regex patterns once
    space_pattern = re.compile(r'\s+')
    char_pattern = re.compile(r'[^a-z0-9 ]')
    
    # Combine operations to minimize string copies
    s = space_pattern.sub(' ', s.lower().strip())
    s = char_pattern.sub('', s)
    return s.replace(' ', '')

def optimized_sampling(testset, samples_per_category, random_state=42):
    """
    Efficiently sample data from different categories.
    
    Args:
        testset: pandas DataFrame containing the data
        samples_per_category: dict mapping categories to number of samples needed
        random_state: random seed for reproducibility
    
    Returns:
        pandas DataFrame with sampled data
    """
    # Pre-compile regex patterns for string cleaning
    space_pattern = re.compile(r'\s+')
    char_pattern = re.compile(r'[^a-z0-9 ]')
    
    # Vectorized string cleaning function
    def clean_strings_vectorized(series):
        # Convert metadata to strings and clean them
        return series.apply(
            lambda x: char_pattern.sub(
                '', 
                space_pattern.sub(
                    ' ', 
                    x[0]['tentang'].lower().strip()
                )
            ).replace(' ', '')
        )
    
    # Clean all categories at once
    testset['cleaned_category'] = clean_strings_vectorized(testset['metadata'])
    
    # Convert samples_per_category keys to cleaned format for matching
    cleaned_samples_dict = {
        clean_string_optimized(k): v 
        for k, v in samples_per_category.items()
    }
    
    # Get unique categories that we need to sample from
    categories_to_sample = set(cleaned_samples_dict.keys()) & set(testset['cleaned_category'].unique())
    
    # Preallocate list with known size
    sampled_dfs = []
    
    # Use numpy's random generator for better performance
    rng = np.random.default_rng(random_state)
    
    for category in categories_to_sample:
        num_samples = cleaned_samples_dict[category]
        # Use boolean indexing
        category_mask = testset['cleaned_category'] == category
        category_questions = testset[category_mask]
        
        samples_to_take = min(num_samples, len(category_questions))
        
        if samples_to_take > 0:
            # Use numpy's choice for sampling instead of pandas sample
            if samples_to_take == len(category_questions):
                sampled_df = category_questions
            else:
                indices = rng.choice(
                    category_questions.index, 
                    size=samples_to_take, 
                    replace=False
                )
                sampled_df = category_questions.loc[indices]
            sampled_dfs.append(sampled_df)
    
    # Single concat operation
    return pd.concat(sampled_dfs) if sampled_dfs else pd.DataFrame()

final_sampled_dataset = optimized_sampling(testset, samples_per_category)
final_sampled_dataset = final_sampled_dataset.drop(columns=['cleaned_category'])
print(len(final_sampled_dataset))
```

```python
import re

def extract_question(text):
    """
    Extracts just the question from a longer text that may include context, explanations,
    and formatting artifacts.
    
    Args:
        text (str): The input text containing a question and potentially other content
        
    Returns:
        str: The cleaned question text
    """
    # First, try to find a quoted question
    quote_patterns = [
        r'"([^"]+?\?)"',  # Double quotes
        r"'([^']+?\?)'",  # Single quotes
        r'"([^"]+?\?)(?!")',  # Double quote at start only
        r"'([^']+?\?)(?!')"   # Single quote at start only
    ]
    
    # Common intro phrases to remove
    intro_patterns = [
        r"Here is (?:the )?(?:rewritten )?question(?:[^:]*)?:\s*",
        r"This (?:revised )?question:\s*",
        r"Question:\s*"
    ]
    
    # Remove common intro phrases
    for pattern in intro_patterns:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE)
    
    # Try each quote pattern
    for pattern in quote_patterns:
        match = re.search(pattern, text)
        if match:
            return match.group(1).strip()
    
    # If no quotes found, try to find the first sentence that ends with a question mark
    match = re.search(r"([^.!?\n]+?\?)", text)
    if match:
        return match.group(1).strip()
    
    # If everything fails, return the first non-empty line that ends with a question mark
    for line in text.split('\n'):
        line = line.strip()
        if line.endswith('?'):
            return line
    
    # If no question mark is found, return the first non-empty line
    for line in text.split('\n'):
        line = line.strip()
        if line:
            return line
    
    return text.strip()

# Apply the function to the dataset
filtered_dataset['cleaned_question'] = filtered_dataset['question'].apply(extract_question)
```
